# **NN4**

```python
import pandas as pd
import numpy as np
from MLP import MLP
np.random.seed(423)
```

## import data

```python
rings_train = pd.read_csv("../dane/classification/rings3-regular-training.csv")
rings_test = pd.read_csv("../dane/classification/rings3-regular-test.csv")
easy_train = pd.read_csv("../dane/classification/easy-training.csv")
easy_test = pd.read_csv("../dane/classification/easy-test.csv")
xor_train = pd.read_csv("../dane/classification/xor3-training.csv")
xor_test = pd.read_csv("../dane/classification/xor3-test.csv")
```

## **rings3-regular** dataset

### softmax

```python
X_train = rings_train.iloc[:, 0:(len(rings_train.iloc[1]) - 1)]
y_train = rings_train.iloc[:, -1]
X_test = rings_test.iloc[:, 0:(len(rings_test.iloc[1]) - 1)]
y_test = rings_test.iloc[:, -1]

mlp = MLP([
    {"input_size": 2, "output_size": 10, "activation": "sigmoid"},
    {"input_size": 10, "output_size": 3, "activation": "softmax"}
], weight_init="xavier", task="classification")

mlp.train(X_train, y_train, 1000, 0.1, batch_size=32, mode="momentum", normalize=True)
mlp.plot_predictions(X_train, y_train, X_test, y_test)
```

### sigmoid

```python
X_train = rings_train.iloc[:, 0:(len(rings_train.iloc[1]) - 1)]
y_train = rings_train.iloc[:, -1]
X_test = rings_test.iloc[:, 0:(len(rings_test.iloc[1]) - 1)]
y_test = rings_test.iloc[:, -1]

mlp = MLP([
    {"input_size": 2, "output_size": 10, "activation": "sigmoid"},
    {"input_size": 10, "output_size": 3, "activation": "sigmoid"}
], weight_init="xavier", task="classification")

mlp.train(X_train, y_train, 1000, 0.1, batch_size=32, mode="momentum", normalize=True)
mlp.plot_predictions(X_train, y_train, X_test, y_test)
```
