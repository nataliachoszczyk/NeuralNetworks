# **NN5**

```python
import pandas as pd
import numpy as np
from MLP import MLP
import matplotlib.pyplot as plt
import seaborn as sns
import winsound
np.random.seed(23)
```

## import data

```python
multimodal_train = pd.read_csv("../dane/regression/multimodal-large-training.csv")
multimodal_test = pd.read_csv("../dane/regression/multimodal-large-test.csv")
steps_train = pd.read_csv("../dane/regression/steps-large-training.csv")
steps_test = pd.read_csv("../dane/regression/steps-large-test.csv")
rings5_train = pd.read_csv("../dane/classification/rings5-regular-training.csv")
rings5_test = pd.read_csv("../dane/classification/rings5-regular-test.csv")
rings3_train = pd.read_csv("../dane/classification/rings3-regular-training.csv")
rings3_test = pd.read_csv("../dane/classification/rings3-regular-test.csv")
```

# tests on **multimodal-large**

```python
X_train = multimodal_train['x'].values.reshape(-1, 1)
y_train = multimodal_train['y'].values.reshape(-1, 1)
X_test = multimodal_test['x'].values.reshape(-1, 1)
y_test = multimodal_test['y'].values.reshape(-1, 1)
```

```python
mse_data = pd.DataFrame(columns=['activation', 'architecture', 'test_loss'])
```

## sigmoid

### 1 -> 35 -> 1

```python
mlp = MLP([
    {"input_size": 1, "output_size": 35, "activation": "sigmoid"},
    {"input_size": 35, "output_size": 1, "activation": "linear"}
], weight_init="uniform")

mlp.train(X_train, y_train, 10000, 0.005, batch_size=120, normalize=False, mode="standard")

mlp.plot_predictions(X_train, y_train, X_test, y_test, normalize=False)
```

```python
mse_data.loc[len(mse_data)] = ["sigmoid", "1,35,1", 1579.01]
```

# chosen: sigmoid, 2 hidden layers & tanh, 1 hidden layer

## steps-large

```python
X_train = steps_train['x'].values.reshape(-1, 1)
y_train = steps_train['y'].values.reshape(-1, 1)
X_test = steps_test['x'].values.reshape(-1, 1)
y_test = steps_test['y'].values.reshape(-1, 1)
```

### sigmoid, 2 hidden layers

```python
mlp = MLP([
    {"input_size": 1, "output_size": 35, "activation": "sigmoid"},
    {"input_size": 35, "output_size": 35, "activation": "sigmoid"},
    {"input_size": 35, "output_size": 1, "activation": "linear"}
], weight_init="uniform")

mlp.train(X_train, y_train, 10000, 0.005, batch_size=120, normalize=False, mode="standard")

mlp.plot_predictions(X_train, y_train, X_test, y_test, normalize=False)
```

### tanh, 1 hidden layer

```python
mlp = MLP([
    {"input_size": 1, "output_size": 35, "activation": "tanh"},
    {"input_size": 35, "output_size": 1, "activation": "linear"}
], weight_init="uniform")

mlp.train(X_train, y_train, 10000, 0.005, batch_size=120, normalize=False, mode="standard")

mlp.plot_predictions(X_train, y_train, X_test, y_test, normalize=False)
```

## rings5-regular

```python
X_train = rings5_train.iloc[:, 0:(len(rings5_train.iloc[1]) - 1)]
y_train = rings5_train.iloc[:, -1]
X_test = rings5_test.iloc[:, 0:(len(rings5_test.iloc[1]) - 1)]
y_test = rings5_test.iloc[:, -1]
```

### sigmoid, 2 hidden layers

```python
mlp = MLP([
    {"input_size": 2, "output_size": 35, "activation": "sigmoid"},
    {"input_size": 35, "output_size": 35, "activation": "sigmoid"},
    {"input_size": 35, "output_size": 5, "activation": "softmax"}
], weight_init="uniform", task="classification")

mlp.train(X_train, y_train, 10000, 0.005, batch_size=120, normalize=False, mode="standard")

mlp.plot_predictions(X_train, y_train, X_test, y_test, normalize=False)
```

### tanh, 1 hidden layer

```python
mlp = MLP([
    {"input_size": 2, "output_size": 35, "activation": "tanh"},
    {"input_size": 35, "output_size": 5, "activation": "softmax"}
], weight_init="uniform", task="classification")

mlp.train(X_train, y_train, 10000, 0.005, batch_size=120, normalize=False, mode="standard")

mlp.plot_predictions(X_train, y_train, X_test, y_test, normalize=False)
```
